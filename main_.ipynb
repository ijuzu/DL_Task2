```python
import json

def load_cities(path):
    cities = set()
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                data = json.loads(line)
                city = data["instruction"].replace("Apibūdink: ", "").strip()
                cities.add(city)
    return cities

train_cities = load_cities("train.jsonl")
test_cities  = load_cities("test.jsonl")

overlap = train_cities.intersection(test_cities)

print(f"Train cities: {len(train_cities)}")
print(f"Test cities:  {len(test_cities)}")
print(f"Overlap:      {len(overlap)}")

if overlap:
    print("\nOverlapping cities:")
    for c in sorted(overlap):
        print("-", c)
else:
    print("\nNo overlap ✅")
```

```python
# optional: detect similar names
for test_city in test_cities:
    for train_city in train_cities:
        if test_city in train_city or train_city in test_city:
            print("Similar:", train_city, "<->", test_city)
```

```python
from difflib import SequenceMatcher

def similar(a, b):
    return SequenceMatcher(None, a, b).ratio() > 0.8
```

```python
clean_test = []

for test_city in test_cities:
    if not any(similar(test_city, train_city) for train_city in train_cities):
        clean_test.append(test_city)

print(f"Truly unseen cities: {len(clean_test)}")

for city in clean_test[:10]:
    print("-", city)
```
